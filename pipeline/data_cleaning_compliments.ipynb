```python
# Obtain the exact product name from 'main' language
def get_main_product_name(prod_name):
    if prod_name is None:
        return None
    for item in prod_name:
        if item['lang'] == 'main':
            return item['text']
    return None
```

## Data Cleaning

### Brands_Tags

```python
# Replace all the inconsistent values ['xx:Compliments', 'compliments'] with a standard value ['xx:compliments']
def normalize_brand_tags(tags):
    normalized_brand_tag = []
    for tag in tags:
        #split the brand_tag by ':' and convert it to lowercase
        brand = tag.split(":", 1)[-1].lower()
        normalized_brand_tag.append(f'xx:{brand}')
    return normalized_brand_tag
```

```python
# Filter out brands_tags which are not ['xx:compliments']
inconsistent_tag = compliments_df['brands_tags'].apply(lambda tags: any(t != f'xx:{t.split(":", 1)[-1].lower()}' for t in tags))

# Apply normalization function to only those tags which are inconsistent
compliments_df.loc[inconsistent_tag, 'brands_tags'] = compliments_df.loc[inconsistent_tag, 'brands_tags'].apply(normalize_brand_tags)
```

```python
# Verify whether the normalization has been applied
compliments_df['brands_tags'].explode().unique()
```
