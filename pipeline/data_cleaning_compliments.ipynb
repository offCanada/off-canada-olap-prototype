```python
# Obtain the exact product name from 'main' language
def get_main_product_name(prod_name):
    if prod_name is None:
        return None
    for item in prod_name:
        if item['lang'] == 'main':
            return item['text']
    return None
```

```python
# Obtain the exact product name from 'main' language where countries_tags is null
null_countries_products = compliments_df.loc[compliments_df['countries_tags'].isnull(), 'product_name'].apply(get_main_product_name)
null_countries_products
```
